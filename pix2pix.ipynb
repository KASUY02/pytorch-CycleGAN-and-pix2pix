```python
import os
import re

# 1. 処理対象のフォルダを指定
raw_dir = './datasets/shadow_synthesis_raw'
subfolders = ['no_shadow', 'mask', 'shadow']

# 半角 (1) も 全角 （1） も両方捉える正規表現パターン
pattern = re.compile(r'[\s\_]*[\(\（]\d+[\)\）](\.[a-zA-Z0-9]+)$')

total_changed = 0

for folder_name in subfolders:
    target_dir = os.path.join(raw_dir, folder_name)
    if not os.path.exists(target_dir):
        print(f"スキップ: フォルダが存在しません -> {target_dir}")
        continue
        
    print(f"--- 【{folder_name}】の処理中 ---")
    files = os.listdir(target_dir)
    changed_in_folder = 0
    
    for filename in files:
        # パターンにマッチするかチェック
        match = pattern.search(filename)
        if match:
            ext = match.group(1) # .png や .jpg を取得
            # カッコ数字部分を消して拡張子だけ残す
            new_filename = pattern.sub(ext, filename)
            
            old_path = os.path.join(target_dir, filename)
            new_path = os.path.join(target_dir, new_filename)
            
            # 同名ファイルとの重複チェック
            if os.path.exists(new_path) and old_path != new_path:
                print(f"  [重複ためスキップ]: {filename} -> (既に {new_filename} が存在)")
            else:
                os.rename(old_path, new_path)
                print(f"  [変更完了]: {filename}  -->  {new_filename}")
                changed_in_folder += 1
                total_changed += 1

    if changed_in_folder == 0:
        print("  対象となるファイルが見つかりませんでした。")
    print()

print(f"【結果】合計 {total_changed} 件のファイル名を変更しました。")
```

```python
import os
from PIL import Image

raw_dir = './datasets/shadow_synthesis_raw'
out_dir = './datasets/shadow_synthesis'

no_shadow_dir = os.path.join(raw_dir, 'no_shadow')
mask_dir = os.path.join(raw_dir, 'mask')
shadow_dir = os.path.join(raw_dir, 'shadow')

# 出力先（A, B）フォルダの作成
os.makedirs(os.path.join(out_dir, 'A/train'), exist_ok=True)
os.makedirs(os.path.join(out_dir, 'B/train'), exist_ok=True)

filenames = [f for f in os.listdir(no_shadow_dir) if f.endswith(('.jpg', '.png', '.jpeg'))]

success_count = 0
for name in filenames:
    path_no_shadow = os.path.join(no_shadow_dir, name)
    path_mask = os.path.join(mask_dir, name)
    path_shadow = os.path.join(shadow_dir, name)

    if not os.path.exists(path_mask) or not os.path.exists(path_shadow):
        print(f"スキップ: {name} に対応する mask または shadow が見つかりません。")
        continue

    # 1. 元画像の読み込み
    img_no_shadow = Image.open(path_no_shadow).convert('RGB')
    img_mask = Image.open(path_mask).convert('RGB')
    img_shadow = Image.open(path_shadow).convert('RGB')

    # 2. 影なし画像のサイズ (幅 W, 高さ H) を基準にする
    w, h = img_no_shadow.size

    # 万が一マスクや影ありのサイズがずれている場合のみ、影なしのオリジナルサイズに合わせる
    if img_mask.size != (w, h):
        img_mask = img_mask.resize((w, h), Image.Resampling.LANCZOS)
    if img_shadow.size != (w, h):
        img_shadow = img_shadow.resize((w, h), Image.Resampling.LANCZOS)

    # 3. A側：影なし + マスク を横に並べて結合 (幅 2W, 高さ H)
    img_A = Image.new('RGB', (w * 2, h))
    img_A.paste(img_no_shadow, (0, 0))
    img_A.paste(img_mask, (w, 0))
    img_A.save(os.path.join(out_dir, 'A/train', name))

    # 4. B側：影あり画像 (幅 W, 高さ H)
    img_shadow.save(os.path.join(out_dir, 'B/train', name))
    success_count += 1

print(f"【結合完了】 {success_count} ペアの画像をオリジナル解像度のまま生成しました！")
```

```python
import os
from PIL import Image, ImageOps

train_dir = './datasets/shadow_synthesis/train'
filenames = [f for f in os.listdir(train_dir) if f.endswith(('.png', '.jpg', '.jpeg'))]

def pad_to_multiple(img, multiple=256):
    w, h = img.size
    # multiple (256) の倍数になる新しい幅と高さを計算
    new_w = ((w + multiple - 1) // multiple) * multiple
    new_h = ((h + multiple - 1) // multiple) * multiple
    
    # 足りない分を黒帯（0）でパディング
    pad_w = new_w - w
    pad_h = new_h - h
    
    # (left, top, right, bottom)
    padding = (0, 0, pad_w, pad_h)
    return ImageOps.expand(img, padding, fill=0)

count = 0
for name in filenames:
    path = os.path.join(train_dir, name)
    img = Image.open(path)
    
    # AとBが横結合された状態（幅が2倍）なので、UNetに適した形にパディング
    w, h = img.size
    # A単体・B単体の幅は w // 2
    single_w = w // 2
    
    # それぞれの単体画像が256の倍数になるように調整
    new_single_w = ((single_w + 256 - 1) // 256) * 256
    new_h = ((h + 256 - 1) // 256) * 256
    
    img_A = img.crop((0, 0, single_w, h))
    img_B = img.crop((single_w, 0, w, h))
    
    pad_A = ImageOps.expand(img_A, (0, 0, new_single_w - single_w, new_h - h), fill=0)
    pad_B = ImageOps.expand(img_B, (0, 0, new_single_w - single_w, new_h - h), fill=0)
    
    # 再度結合
    img_padded = Image.new('RGB', (new_single_w * 2, new_h))
    img_padded.paste(pad_A, (0, 0))
    img_padded.paste(pad_B, (new_single_w, 0))
    
    img_padded.save(path)
    count += 1

print(f"【パディング完了】 {count} 枚の画像を256の倍数サイズに調整しました。")
```
